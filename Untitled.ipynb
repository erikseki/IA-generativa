```python
import re
import unicodedata

# Funções necessárias do notebook
def limpar_texto_completo(texto):
    """
    Limpa o texto, removendo URLs, menções, hashtags, caracteres especiais,
    pontuações, números, espaços extras e acentos, convertendo para lowercase.
    """
    if isinstance(texto, str):
        # Remove URLs
        texto = re.sub(r'http\S+|www\S+|https\S+', '', texto, flags=re.MULTILINE)
        # Remove menções (@)
        texto = re.sub(r'@\w+', '', texto)
        # Remove hashtags (#)
        texto = re.sub(r'#\w+', '', texto)
        # Remove caracteres especiais e pontuações, mantendo espaços
        texto = re.sub(r'[^\w\s]', '', texto)
        # Remove números
        texto = re.sub(r'\d+', '', texto)
        # Remove acentos
        texto = ''.join((c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn'))
        # Remove espaços extras
        texto = ' '.join(texto.split())
        return texto.lower()
    return ''
```

```python
# testando o tratamento de texto 
texto = "Amei o produto! 😍 Super recomendo! #ótimo @loja http://site.com"
print(limpar_texto_completo(texto))
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('portuguese'))
stop_words
```

```python
def remover_stop_words(texto):
    """
    Remove stop words de um texto
    """
    if isinstance(texto, str):
        stop_words = set(stopwords.words('portuguese'))
        stop_words.remove('não')
        palavras = texto.split()
        texto_sem_stopwords = [palavra for palavra in palavras if palavra not in stop_words]
        return ' '.join(texto_sem_stopwords)
    return ''
```

```python
# aplicando a limpeza das stop words
df['comentario_limpo_stop_words'] = df['Comentario'].apply(remover_stop_words)
```

```python
# aplicando limpeza no texto completo
df['comentario_limpo'] = df['comentario_limpo_stop_words'].apply(limpar_texto_completo)
```
